```python
area_summary["pollution_normalized"] = (
    (area_summary["pollution_index"] - area_summary["pollution_index"].min())
    /
    (area_summary["pollution_index"].max() - area_summary["pollution_index"].min())
)
```

```python
area_summary["population_normalized"] = (
    (area_summary["population"] - area_summary["population"].min())
    /
    (area_summary["population"].max() - area_summary["population"].min())
)
```

```python
area_summary["crime_rank"] = (
    area_summary["crime_rate"]
    .rank(method="dense", ascending=False)
    .astype(int)
)

area_summary[
    ["crime_rank", "area", "crime_rate"]
].sort_values("crime_rank").head(10)
```

```python
pollution_Q1 = area_summary["pollution_index"].quantile(0.25)
pollution_Q3 = area_summary["pollution_index"].quantile(0.75)

def classify_pollution(index):
    if index <= pollution_Q1:
        return "Low"
    elif index <= pollution_Q3:
        return "Medium"
    else:
        return "High"

area_summary["pollution_category"] = (
    area_summary["pollution_index"].apply(classify_pollution)
)

area_summary[["area", "pollution_index", "pollution_category"]]
```

```python
pollution_crime = (
    area_summary.groupby("pollution_category")["crime_rate"]
    .mean()
    .sort_values(ascending=False)
)

pollution_crime
```

## Key Findings

1. Civil Lines has the highest average crime rate among the analyzed areas.
2. Preet Vihar has the lowest average crime rate.
3. Most analyzed areas fall under the Medium crime category.
4. The correlation between average income and crime rate is almost zero, indicating no meaningful linear relationship in this dataset.
5. Population density also shows only a weak relationship with crime rate.
6. Pollution index has a weak negative correlation with crime rate; therefore, higher pollution does not necessarily mean higher crime in this dataset.
7. High-income areas have a higher average crime rate than medium- and low-income areas, but this comparison does not establish causation.
8. Overall, no single numerical factor strongly explains the variation in crime rate across the analyzed Delhi areas.

## Conclusion

This project analyzed different areas of Delhi using population, average income, crime rate, pollution index, and population density.

The analysis shows that crime rate varies considerably across different areas. However, the correlation analysis indicates that no single factor among income, population, pollution, or population density has a strong linear relationship with crime rate in this dataset.

The project demonstrates how Python, NumPy, Pandas, and Matplotlib can be used to clean, analyze, visualize, and extract meaningful insights from real-world style data.

Overall, the analysis provides a data-driven overview of Delhi's area-wise characteristics and highlights that crime patterns are influenced by multiple factors rather than one simple variable.
